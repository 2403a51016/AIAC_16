Write a python function that converts a given input from centimeters to inches
Example: 20 cm = 7.87 inches


```python
def cm_to_inches(cm):
  """Converts centimeters to inches."""
  return cm * 0.393701

# Example usage:
cm_value = 20
inches_value = cm_to_inches(cm_value)
print(f"{cm_value} cm = {inches_value:.2f} inches")
```
