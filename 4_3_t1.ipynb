task 1 prompt: Write a python function which takes a year as an input and checks whether the year is a leap year


```python
def is_leap(year):
  """Checks if a year is a leap year."""
  return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

# Example usage
print(is_leap(2000))
print(is_leap(1900))
print(is_leap(2024))
```
