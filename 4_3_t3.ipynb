```python
def format_names(names):
  """
  Formats a list of names as first name and last name.

  Args:
    names: A list of strings, where each string is a full name.

  Returns:
    None. Prints the formatted names.
  """
  for name in names:
    parts = name.split()
    if len(parts) >= 2:
      first_name = parts[0]
      last_name = parts[-1]
      print(f"{first_name}, {last_name}")
    else:
      print(f"{name}: Could not format name (requires at least two parts).")

# Example usage:
name_list = ["Narendra Modi", "Rahul Gandhi", "SingleName"]
format_names(name_list)
```
